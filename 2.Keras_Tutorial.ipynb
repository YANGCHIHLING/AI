# Keras深度學習初體驗1
利用keras套件，產生一個32個神經元的深度學習模型進行訓練


```
# 1. 匯入所需的函式庫
# 匯入TensorFlow中的Keras深度學習工具
from tensorflow import keras

# 匯入Sequential(順序)模型，用來依序建立神經網路
from keras.models import Sequential

# 匯入Dense(緻密)全連接層，每個神經元都會接收前一層的輸出
from keras.layers import Dense

# 匯入NumPy，用來進行數值運算及陣列處理
import numpy as np


# 2. 建立訓練資料(模型訓練時用來對照的正確答案)
# 隨機產生1000筆資料，每筆資料包含十個介於0到1之間的浮點數
# 1000代表資料筆數，10代表每筆資料的特徵數量(即每次輸入的資料都有10個特徵)
data = np.random.random((1000, 10))

# 隨機產生1000筆資料，每筆資料只有0或1兩種可能
# 2代表隨機產生的整數範圍為0到1（不包含2）
# size=(1000, 1)代表產生1000列、每列1個數值
labels = np.random.randint(2, size=(1000, 1))


# 3. 查看資料的形狀及內容
# 印出data的形狀，結果為(1000, 10)，表示共有1000筆資料，每筆資料有10個特徵
print(np.shape(data))

# 印出data的第1筆資料（索引從0開始），會顯示該筆資料中的10個浮點數
print(data[0])

# 印出labels的形狀，結果為(1000, 1)，表示共有1000筆標籤資料，每筆只有1個標籤
print(np.shape(labels))

# 印出labels的第1筆資料，結果只會是0或1
print(labels[0])


# 4. 建立神經網路模型
# 建立Sequential模型，神經網路中各層會按照加入的順序依序排列
model = Sequential()

# 加入第1層神經網路（隱藏層）
# (1.)Dense(32)：設定這一層有32個神經元
# (2.)activation='relu'：使用ReLU啟動函數，將負數轉換為0，正數則保持原值
# (3.)input_dim=10：代表輸入資料有10個特徵
model.add(Dense(32, activation='relu', input_dim=10))

# 加入第2層神經網路（輸出層）
# (1.)Dense(1)：設定輸出層只有1個神經元，即最後一層
# (2.)activation='sigmoid'：使用Sigmoid啟動函數，將輸出轉換為0到1之間的數值
# 常用於二元分類問題，模型用於預測資料屬於某一類別的機率
model.add(Dense(1, activation='sigmoid'))


# 5. 編譯神經網路模型
# 設定模型訓練時所使用的優化器、損失函數及評估指標
model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])


# 6. 訓練神經網路模型
# 將訓練資料data及對應標籤labels輸入模型進行訓練
# epochs=10：將全部訓練資料完整學習10次
# batch_size=32：每次取32筆資料進行一個批次的訓練
model.fit(data, labels, epochs=10, batch_size=32)


# 7. 評估模型的表現
# 使用data及labels評估模型
# evaluate會回傳損失值及評估指標
score=model.evaluate(data, labels)

# 印出score[1]，也就是模型的準確率（accuracy）
# score[0]代表損失值（loss）
# score[1]代表準確率（accuracy）
print(score[1])
```
